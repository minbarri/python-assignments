```python
# bmi_calculator.py
# 작성자: 민서
# 기능: BMI 계산기 (입력 + 출력 + 조건문 활용)

# 사용자 입력
name = input("이름을 입력하세요: ")
height_cm = float(input("키를 입력하세요 (cm): "))
weight_kg = float(input("몸무게를 입력하세요 (kg): "))

# 키(m) 단위로 변환
height_m = height_cm / 100

# BMI 계산
bmi = weight_kg / (height_m ** 2)

# 결과 출력
print("-" * 40)
print(f"{name}님의 BMI는 {bmi:.2f}입니다.")

# 조건문으로 판정 출력
if bmi < 18.5:
    print("판정: 저체중 😢")
elif bmi < 23:
    print("판정: 정상 😄")
elif bmi < 25:
    print("판정: 과체중 😐")
else:
    print("판정: 비만 😥")

print("-" * 40)
```
