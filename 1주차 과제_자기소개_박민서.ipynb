```python
# 자기소개 프로그램 (Google Colab용)

def introduce():
    print("\n===================================")
    print("         📝 자기소개 입력 📝")
    print("===================================")

    name = input("이름을 입력하세요: ")
    major = input("전공을 입력하세요: ")
    interest = input("관심 있는 분야를 입력하세요: ")
    hobby = input("취미를 입력하세요: ")
    goal = input("앞으로의 목표를 입력하세요: ")

    print("\n===================================")
    print("           ✨ 자기소개 ✨")
    print("===================================")
    print(f"안녕하세요! 저는 {name}입니다 :)")
    print(f"전공: {major}")
    print(f"관심 분야: {interest}")
    print(f"취미: {hobby}")
    print(f"목표: {goal}")
    print("===================================")

# 메인 루프
while True:
    introduce()

    again = input("\n자기소개를 다시 입력하시겠습니까? (y/n): ").strip().lower()
    if again == 'n':
        print("\n프로그램을 종료합니다. 감사합니다!")
        break
    elif again != 'y':
        print("⚠️ 잘못된 입력입니다. 프로그램을 종료합니다.")
        break
```
